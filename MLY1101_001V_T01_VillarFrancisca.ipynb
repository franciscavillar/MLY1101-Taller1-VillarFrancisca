# Fuente 2 — API 2

* **Nombre de la API:** Digi-API
* **Documentación:** https://digi-api.com/
* **Endpoint utilizado:** https://digi-api.com/api/v1/digimon  
* **Descripción de los datos:** Información de Digimon disponibles en la API, incluyendo datos asociados a cada personaje digimon.   
* **Relación con el objetivo:** Permite recopilar información de personajes pertenecientes a serie Digimon y generar un segundo dataset independiente.  


````python
# CONFIGURACIÓN API 2
API2_URL = 'https://digi-api.com/api/v1/digimon'
API2_MIN_REGISTROS = 200
api2_headers = {}
api2_params = {
    'page': 0,
    'pageSize': 200
}
````

````python
# CONSUMO API 2
api2_registros = []

response = requests.get(API2_URL, headers=api2_headers, params=api2_params, timeout=30)
print('Status API 2:', response.status_code)
response.raise_for_status()

# TODO: extraiga la lista de registros.
# api2_registros = data['results']

data = response.json()

api2_registros = data['content']

print('Registros API 2:', len(api2_registros))
````

````python
# GUARDAR DATASET API 2
API2_ARCHIVO = OUTPUT_DIR / 'dataset_api_2.json'

with open(API2_ARCHIVO, 'w', encoding='utf-8') as f:
    json.dump(api2_registros, f, ensure_ascii=False, indent=2)

print('Archivo generado:', API2_ARCHIVO)
print('Registros guardados:', len(api2_registros))
````

# Fuente 3 — API 3

* **Nombre de la API:** Rick and Morty API
* **Documentación:** https://rickandmortyapi.com/documentation
* **Endpoint utilizado:** https://rickandmortyapi.com/api/character
* **Descripción de los datos:** Información de personajes de la serie Rick and Morty, incluyendo nombre, especie, estado, género y origen.
* **Relación con el objetivo:** Permite recopilar información de personajes pertenecientes a una tercera serie y generar un dataset independiente.


````python
# CONFIGURACIÓN API 3
API3_URL = 'https://rickandmortyapi.com/api/character'
API3_MIN_REGISTROS = 200
api3_headers = {}
api3_params = {
    'page': 1
}
````

````python
# PAGINACIÓN Y CONSUMO API 3

api3_registros = []
page = 1

while len(api3_registros) < API3_MIN_REGISTROS:

    params = {'page': page}

    response = requests.get(
        API3_URL,
        headers=api3_headers,
        params=params,
        timeout=30
    )

    response.raise_for_status()
    data = response.json()

    nuevos = data['results']

    if not nuevos:
        break

    api3_registros.extend(nuevos)

    print('Página:', page, '- Registros acumulados:', len(api3_registros))

    page += 1

api3_registros = api3_registros[:API3_MIN_REGISTROS]

print('Registros API 3:', len(api3_registros))
````

````python
# GUARDAR DATASET API 3
API3_ARCHIVO = OUTPUT_DIR / 'dataset_api_3.json'

with open(API3_ARCHIVO, 'w', encoding='utf-8') as f:
    json.dump(api3_registros, f, ensure_ascii=False, indent=2)

print('Archivo generado:', API3_ARCHIVO)
print('Registros guardados:', len(api3_registros))
````
